# Baseline replication analysis

Loads a run written by `main.run_baseline_experiment` (see `logs/baseline/`) and reports the paper's IMR / RBLM / GL / BE metrics (Kuo, Hsueh & Tsai 2023, Section 2.2).

```python
import matplotlib.pyplot as plt
import paper_metrics

# Set this to the run_id you want to analyze -- matches the prefix of the files in
# logs/baseline/, e.g. "gpt-3_5-turbo_1735000000"
RUN_ID = "REPLACE_WITH_RUN_ID"

moves_df, games_df, config = paper_metrics.load_run(RUN_ID, log_dir="../logs/baseline")
config
```

```python
summary = paper_metrics.summarize_run(moves_df, games_df)
summary
```

```python
imr = paper_metrics.imr_series(moves_df)
imr_by_move = imr.groupby("move_index")["imr"].mean()

plt.figure()
imr_by_move.plot()
plt.xlabel("Move (t)")
plt.ylabel("Mean IMR")
plt.title(f"Average IMR by move -- {RUN_ID}")
plt.show()
```

```python
rblm = paper_metrics.rblm_series(moves_df)
rblm_by_move = rblm.groupby("move_index")["rblm"].mean()

plt.figure()
rblm_by_move.plot()
plt.xlabel("Move (t)")
plt.ylabel("Mean RBLM")
plt.title(f"Average RBLM by move -- {RUN_ID}")
plt.show()
```

```python
plt.figure()
paper_metrics.game_lengths(games_df, unit="fullmoves").plot.hist(bins=20)
plt.xlabel("Game length (fullmoves)")
plt.title(f"Game length distribution -- {RUN_ID}")
plt.show()
```
